# Pandas Basics

Importing pandas

```python
import pandas as pd
```

Loading a csv

```python
df = pd.read_csv('Data/gym.csv')
```

Printing first 5 rows

```python
df.head(5)
```

Printing the last 5 rows

```python
df.tail(5)
```

Printing the column names

```python
df.columns
```

Show basic information about the dataset (Columns, non-null counts, data types)

```python
df.info()
```

Show basic descriptive statistics for numerical columns

```python
df.describe()
```

Show shape of the dataset

```python
df.shape
```

Count the number of missing values in each column

```python
df.isnull().sum()
```

Drop any rows that have missing values

```python
df.dropna(inplace=True)
```

Drop columns that have missing values

```python
df.dropna(inplace=True, axis=1)
```

Fill missing values in a column with the mean of that column

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

Select a single column form the dataset

```python
df['age']
```

Selecting multiple columns

```python
df[['age','gender']]
```

Select rows where age is > 30

```python
df[df['age']>30]
```

Select rows where age is > 30 and gender is male

```python
df[(df['age']>30) & (df['gender'] =='Male')]
```

Select rows where gender is either 'Male' or 'Other'

```python
df[df['gender'].isin(['Male','Other'])]
```

Group by gender and get the average age for each gender

```python
df.groupby(['gender'])['age'].mean()
```

Group by gender and membership type and then get the average age

```python
df.groupby(['gender','membership_type'])['age'].mean()
```

Making the result of the groupby method a data frame

```python
df_new = df.groupby(['gender','membership_type'])['age'].mean().reset_index()
df_new
```

Sort the data frame by the average age in descending order

```python
df_new.sort_values('age', ascending=False)
```

Create a new column based on existing columns

```python
df['age_plus_5'] = df['age'] + 5
df.head(1)
```

Create a new column by applying a function to an existing column

```python
df['age_squared'] = df['age'].apply(lambda x: x**2)
df.head(1)
```

Rename the column age_plus_5 to age_added_5

```python
df.rename(columns={'age_plus_5' : 'age_added_5'},inplace=True)
df.head(1)
```

Convert the age column to float

```python
df['age'] = df['age'].astype(float)
df['age'].head(1)
```

Create two new dataframes that share the member_id column

```python
df_1 = df[['member_id','age','gender']]
df_2 = df[['member_id','membership_type','check_in_time']]
```

```python
df_1.head()
```

```python
df_2.head()
```

Merge the two datasets on member_id (left join)

```python
df_merged = pd.merge(df_1,df_2, on='member_id', how='left')
```

```python
df_merged.head()
```

Import matplotlib

```python
import matplotlib.pyplot as plt
```

Plot age on the y-axis against member_id on the x-axis

```python
y = df['age']
x = df['member_id']
y1 = df['age_added_5']

fig, ax = plt.subplots(figsize=(10,6))
ax.set_title('Age vs Member ID')
ax.set_xlabel('Member ID')
ax.set_ylabel('Age')
ax.plot(x,y,color='blue',linestyle='-',label='Age')
ax.plot(x,y1,color='red',linestyle='--',label='Age + 5')
ax.legend()
plt.show()
```

Create a scatter plot of age vs member ID

```python
fig, ax = plt.subplots(figsize=(10,6))
ax.scatter(x,y,marker='o',color='green',label='Age points')
ax.grid(True)
plt.show()
```
